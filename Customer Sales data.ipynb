#### **- Date Column**

##### **a. Last_Purchase_Date**

```python
# Convert date columns temporarily into datetime format

signup_dates = pd.to_datetime(df_clean['Signup_Date'])

last_purchase_dates = pd.to_datetime(
    df_clean['Last_Purchase_Date'],
    errors='coerce'
)

# Identify invalid dates occurring before signup dates

invalid_dates = (
    last_purchase_dates.notna() &
    (last_purchase_dates < signup_dates)
)

# Identify missing and invalid dates

missing_or_invalid = (
    last_purchase_dates.isna() | invalid_dates
)

# Calculate median using valid purchase dates

date_median = last_purchase_dates[
    ~missing_or_invalid
].median()

# Prepare median dates for each customer

median_dates = pd.Series(
    date_median,
    index=df_clean.index
)

# Ensure replacement dates are not earlier than signup dates

replacement_dates = pd.concat(
    [signup_dates, median_dates],
    axis=1
).max(axis=1)

# Replace missing and invalid dates

last_purchase_dates.loc[missing_or_invalid] = (
    replacement_dates.loc[missing_or_invalid]
)

# Update the original column

df_clean['Last_Purchase_Date'] = last_purchase_dates
```

**Inference:**
The Last_Purchase_Date column was **temporarily converted into datetime format** to identify **missing and invalid dates**. A total of **1,012 missing dates and 936 invalid dates were identified**. These values were filled using the median date **(18-Oct-2025)**, while the Signup_Date was used whenever the median date preceded registration. Finally, the column was converted back to string format until the permanent data type conversion stage.

#### **2.8.1.3 Identifying Outliers Using the IQR Method**

**Purpose:** To identify potential outliers by calculating the Interquartile Range (IQR), which measures the spread of the middle 50% of observations.

The IQR is calculated as:
\[
IQR = Q3-Q1
\]

The lower and upper boundaries are determined using:

\[
Lower=Q1-1.5(IQR)
\]

\[
Upper=Q3+1.5(IQR)
\]

```python
numerical_cols = [
    'Age',
    'Purchase_Amount',
    'Feedback_Score',
    'Purchase_Duration_Days',
    'Days_Since_Last_Purchase',
    'Purchase_Amount_Difference_From_City_Median'
]

for column in numerical_cols:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    print(f'{column}: {len(outliers)} potential outliers')
```
