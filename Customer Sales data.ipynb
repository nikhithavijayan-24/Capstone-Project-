## **2.2 Text Standardization and Inconsistency Correction**

Text standardization is the process of improving consistency across textual data by correcting inconsistent formats and removing unwanted spaces, characters, and words. ### **2.2.2 Handling Inconsistencies**

#### **a. Treating Age Column**

```python
df_clean['Age'].unique()
```

```python
df_clean['Age']=df_clean['Age'].str.strip().str.replace('years',' ',case=False,regex=False).str.strip()
df_clean['Age'].unique()
```

The Age column contained inconsistent text formats and unnecessary spaces. The **year** text was removed using **case=False** to match both uppercase and lowercase variations, while **regex=False** treats it as ordinary text rather than a regular expression. The values are now standardized, while the final data type conversion is reserved for a later stage.

#### **b. Treating Gender Column**

```python
df_clean['Gender'].unique()
```

```python
df_clean['Gender']=df_clean['Gender'].str.strip().str.title().replace({
                                                                    'M':'Male','F':'Female'})

df_clean['Gender'].unique()
```

The Gender column contained inconsistent representations such as **m, MALE, f, and FEMALE**. Unnecessary spaces were removed using **str.strip()**, and letter cases were standardized using **str.title()**. The abbreviations **M** and **F** were replaced with Male and Female to ensure consistent category labels.

#### **c. Treating City Column**

```python
df_clean['City'].unique()
```

```python
df_clean['City']=df_clean['City'].str.strip().str.title()
df_clean['City'].unique()
```

The City column contained inconsistent capitalization and unnecessary spaces. These were corrected using **str.strip()** and **str.title()** to maintain a uniform format across city names.

#### **d. Treating Country Column**

```python
df_clean['Country'].unique()
```

```python
df_clean['Country']=df_clean['Country'].str.strip().str.title().replace({'Ind':'India'})
df_clean['Country'].unique()
```

The Country column contained inconsistent representations such as **IND`, india**, and **InDia**. **Whitespace** and **capitalization** were standardized, and the abbreviation **Ind** was replaced with **India** to maintain a consistent country name.

## **2.3 Missing Values / Null Values**

Missing values are one of the major data quality issues identified in your dataset. Before handling them, we need to examine their distribution across columns

### **2.3.1 Handling Null values**

**Involves**

#### **2.3.1.1 Identifying the missing Values**

The **isna/isnull().sum()** function identifies and counts the missing values in each column. It helps determine which attributes contain incomplete records and the extent of missing data that needs to be addressed during preprocessing.

```python
for col in df_clean.columns:
    if df_clean[col].isnull().sum() > 0:
        print(f"\n{col}")
        print(df_clean[col].value_counts(dropna=False))
```

#### **2.3.1.2 Filling Missing Values**

#### **- Numerical Columns**

##### **a. Age**

```python
df_clean['Age'] = df_clean['Age'].replace('nan', pd.NA)
```

```python
age_median = pd.to_numeric(df_clean['Age'], errors='coerce').median()
df_clean['Age'] = df_clean['Age'].fillna(str(age_median))
print(age_median)
```

The text value **nan** was first converted into an actual missing value. The Age values were temporarily converted into numeric format using **errors='coerce** to handle invalid entries, and the calculated median (**43.0**) was stored in **age_median** for reuse. The median was then converted back into a string using **str()** to fill the missing values while preserving the column's original data type until the final conversion stage.

##### **b. Purchase_Amount**

```python
print(df_clean['Purchase_Amount'].median())
df_clean['Purchase_Amount']=df_clean['Purchase_Amount'].fillna(df_clean['Purchase_Amount'].median())
print(df_clean['Purchase_Amount'])
```

The median value is **24330.0**. Missing values in the Purchase_Amount column were filled using the **median** because the dataset contains **extreme purchase values** that could influence the mean. The median provides a more **reliable central value** and is less affected by such extremes

##### **c. Feedback_Score**

```python
print(df_clean['Feedback_Score'].median())
df_clean['Feedback_Score']=df_clean['Feedback_Score'].fillna(df_clean['Feedback_Score'].median())
```

The median is **5.0**. Missing Feedback Score values were filled using the **median** to maintain the central tendency of customer ratings while reducing the influence of unusually high or low scores.

#### **- Categorical Columns**

##### **a. Gender**

```python
print(df_clean['Gender'].mode()[0])
df_clean['Gender'] = df_clean['Gender'].fillna(df_clean['Gender'].mode()[0])
```

Missing values in the Gender column were filled using the **mode** which is **Female**, which represents the most frequently occurring category. This method was chosen because **Gender** is a **categorical column**, and using the most common category helps maintain consistency in the data.

##### **b. City**

```python
print(df_clean['City'].mode()[0])
df_clean['City'] = df_clean['City'].fillna(df_clean['City'].mode()[0])
```

Missing City values were replaced with the **most frequently occurring city** that is **Mumbai** to maintain consistency in the categorical data.

##### **c. Country**

```python
df_clean['Country'] = df_clean['Country'].fillna('India')
```

Since **India** is the only country represented in the dataset, all missing Country values were filled with **India** to maintain consistency and completeness.

##### **d. Customer ID**

```python
missing_id = df_clean['Customer_ID'].isnull()
df_clean.loc[missing_id, 'Customer_ID'] = [
    f'UNKNOWN_{i+1}' for i in range(missing_id.sum())
]
print(df_clean['Customer_ID'])
```

Missing Customer_ID values were replaced with **unique placeholder identifiers** to avoid assigning the **same ID to multiple records**. These generated identifiers help maintain record identification without treating them as actual customer IDs.

#### **- Date Column**

##### **a. Last_Purchase_Date**

```python
# Temporarily convert to datetime
df_clean['Last_Purchase_Date'] = pd.to_datetime(
    df_clean['Last_Purchase_Date'],
    errors='coerce'
)

# Fill missing values with a fixed date
df_clean['Last_Purchase_Date'] = (
    df_clean['Last_Purchase_Date'].fillna('2025-10-08')
)

# Convert back to string
df_clean['Last_Purchase_Date'] = (
    df_clean['Last_Purchase_Date'].dt.strftime('%Y-%m-%d')
)
df_clean['Last_Purchase_Date']
```

The Last_Purchase_Date column was **temporarily converted into datetime format** to identify **missing and invalid dates**. A total of **1,012 missing dates and 936 invalid dates were identified**. Finally, the column was converted back to string format until the permanent data type conversion stage.

### **2.7.1 Numerical Columns**


##### **a. Purchase_Duration_Days**

To calculate the number of days between customer registration and their last recorded purchase, helping understand the time interval between signup and purchase.

```python
df_clean['Purchase_Duration_Days'] = (
    df_clean['Last_Purchase_Date'] - df_clean['Signup_Date']
).dt.days
```

The difference between **Last_Purchase_Date and Signup_Date** was calculated to determine the number of days between customer registration and their last recorded purchase. The **.dt.days accessor converts this date difference into a numerical value**. For records with imputed purchase dates, the calculated duration represents an estimate rather than an actual recorded interval.

```python
(df_clean['Purchase_Duration_Days'] < 0).sum()
```

##### **b. Days_Since_Last_Purchase**

To calculate how many days have elapsed since each customer's last recorded purchase, helping understand purchase recency.

```python
reference_date = df_clean['Last_Purchase_Date'].max()

df_clean['Days_Since_Last_Purchase'] = (
    reference_date - df_clean['Last_Purchase_Date']
).dt.days
```

The latest available purchase date in the dataset was selected as a **common reference date using .max()**. The difference between this reference date and each recorded purchase date was calculated in days using **.dt.days**. Lower values indicate more recent purchases relative to the latest date in the dataset. Results for imputed dates are estimates.

##### **c. Purchase_Amount_Percentile**

To identify the relative position of each purchase amount compared with other purchase records in the dataset.

```python
df_clean['Purchase_Amount_Percentile'] = (
    df_clean['Purchase_Amount'].rank(pct=True) * 100
)
```

The **.rank(pct=True)** function calculates the **relative percentile position** of **each purchase amount, and multiplying by 100** expresses it as a percentage. This feature helps compare **purchase amounts** based on their **relative position** within the dataset.

##### **d. Purchase_Amount_Difference_From_City_Median**

To compare each purchase amount with the median purchase amount of its respective city.

```python
df_clean['Purchase_Amount_Difference_From_City_Median'] = (
    df_clean['Purchase_Amount']
    - df_clean.groupby('City')['Purchase_Amount'].transform('median')
)
```

The **groupby()** function groups records by **city**, while transform('median') calculates the **median purchase amount for each city** and aligns it with the corresponding records. **Subtracting** this value from **each purchase amount** helps **identify whether an individual purchase is above or below its city's median**.

### **2.7.2 Categorical Columns**


##### **a. Age_Group**

To classify customers into different age categories and understand the distribution of customers across age groups.

```python

df_clean['Age_Group'] = pd.cut(
    df_clean['Age'],
    bins=[0, 17, 25, 35, 45, 60, 100],
    labels=['Under 18', '18-25', '26-35', '36-45', '46-60', 'Above 60']
)
```

The **pd.cut()** function was used to **divide numerical age values into six predefined intervals**. **Under 18, 18–25, 26–35, 36–45, 46–60, and Above 60**. The bins parameter defines the age boundaries, while **labels** assigns **meaningful category names to each interval**, making demographic analysis easier.



##### **b. Customer_Satisfaction_Level**

To classify customers into different satisfaction categories based on their feedback scores.

```python
df_clean['Customer_Satisfaction_Level'] = pd.cut(
    df_clean['Feedback_Score'],
    bins=[0, 3, 6, 8, 10],
    labels=['Dissatisfied', 'Neutral', 'Satisfied', 'Highly Satisfied']
)
```

The **pd.cut()** function was used to **divide feedback scores into four satisfaction categories** using **predefined score intervals**. The **bins** parameter specifies the **score ranges**, while **labels** assigns **meaningful satisfaction levels** to make customer feedback easier to interpret.


##### **c. Customer_Spending_Segment**
To classify purchase records into different spending categories based on their purchase amounts.

```python
df_clean['Customer_Spending_Segment'] = pd.qcut(
    df_clean['Purchase_Amount'],
    q=4,
    labels=['Low', 'Moderate', 'High', 'Top']
)
```

The **pd.qcut()** function was used to **divide purchase amounts into four quantile-based spending segments**. The **q=4 parameter creates four groups containing approximately equal numbers of records,** allowing comparisons across different spending levels.

#### **2.8.1.3 Identifying Outliers Using the IQR Method**

To identify potential outliers by calculating the Interquartile Range (IQR), which measures the spread of the middle 50% of observations.

The IQR is calculated as:
\[
IQR = Q3-Q1
\]

The lower and upper boundaries are determined using:

\[
Lower=Q1-1.5(IQR)
\]

\[
Upper=Q3+1.5(IQR)
\]

```python
numerical_cols = [
    'Age',
    'Purchase_Amount',
    'Feedback_Score',
    'Purchase_Duration_Days',
    'Days_Since_Last_Purchase',
    'Purchase_Amount_Difference_From_City_Median'
]

for column in numerical_cols:
    Q1 = df_clean[column].quantile(0.25)
    Q3 = df_clean[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df_clean[
        (df_clean[column] < lower_limit) |
        (df_clean[column] > upper_limit)
    ]

    print(f'{column}: {len(outliers)} potential outliers')
```

The IQR method identifies potential outliers by comparing numerical observations against the calculated lower and upper boundaries. The 1.5 multiplier is used to establish these limits, allowing unusual values in different numerical columns to be identified separately

#### **2.8.1.4 Inspecting Potential Outliers**

##### **a. Inspecting Potential Purchase_Amount Outliers**
To examine the actual records identified as potential outliers before deciding on an appropriate treatment method.

### **2.8.2 Handling Outliers**
After identifying potential outliers, appropriate treatment is required to improve the quality and reliability of the dataset. In this project, the IQR method is used to remove observations that fall outside the calculated lower and upper boundaries. The filtered dataset retains observations within the defined limits for further analysis.

#### **2.8.2.1 Removing Outliers from Purchase Amount**
To remove unusually high purchase amounts identified as potential outliers, as these values may represent data-entry errors and can significantly affect statistical analysis.

```python
Q1 = df_clean['Purchase_Amount'].quantile(0.25)
Q3 = df_clean['Purchase_Amount'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_clean = df_clean[
    (df_clean['Purchase_Amount'] >= 0) &
    (df_clean['Purchase_Amount'] >= lower_bound) &
    (df_clean['Purchase_Amount'] <= upper_bound)
].copy()
```

The IQR method identified and removed four unusually **high purchase amounts of 9,999,999.0.** Additionally, **four negative purchase values of -500.0 were excluded** using the non-negative condition, while zero-valued records were retained.

#### **2.8.2.2 Removing Outliers from Age**
To remove invalid age observations that fall outside the calculated IQR boundaries and maintain reasonable age values in the dataset.

```python
Q1 = df_clean['Age'].quantile(0.25)
Q3 = df_clean['Age'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

df_clean = df_clean[
    (df_clean['Age'] >= lower_bound) &
    (df_clean['Age'] <= upper_bound)
].copy()
```

The IQR method was applied to filter out unusual age observations, including invalid values such as **-10 and 250**. Removing these records helps maintain a more reliable age distribution for demographic analysis.

#### **2.8.2.3 Purchase Amount Difference From City Median**




The IQR method identified **four potential outliers** in the **Purchase_Amount_Difference_From_City_Median column**. These were the same records containing unusually **high purchase amounts of 9,999,999.0,** which were removed during **Purchase_Amount outlier treatment.** Therefore, no separate removal was required for this derived column.

### **2.8.3 Updating Affected Categorical Columns After Outlier Removal**

```python

# ==========================================
# 1. NUMERICAL COLUMNS
# ==========================================

# Recalculate Purchase Duration
df_clean['Purchase_Duration_Days'] = (
    df_clean['Last_Purchase_Date'] -
    df_clean['Signup_Date']
).dt.days

# Recalculate Days Since Last Purchase
reference_date = df_clean['Last_Purchase_Date'].max()

df_clean['Days_Since_Last_Purchase'] = (
    reference_date - df_clean['Last_Purchase_Date']
).dt.days

# Recalculate Purchase Amount Percentile
df_clean['Purchase_Amount_Percentile'] = (
    df_clean['Purchase_Amount'].rank(pct=True) * 100
)

# Recalculate Purchase Amount Difference from City Median
df_clean['Purchase_Amount_Difference_From_City_Median'] = (
    df_clean['Purchase_Amount'] -
    df_clean.groupby('City')['Purchase_Amount'].transform('median')
)


# ==========================================
# 2. CATEGORICAL COLUMNS
# ==========================================

# Recreate Age Group
df_clean['Age_Group'] = pd.cut(
    df_clean['Age'],
    bins=[0, 17, 25, 35, 45, 60, 100],
    labels=['Under 18', '18-25', '26-35', '36-45', '46-60', 'Above 60']
)

# Recreate Customer Satisfaction Level
df_clean['Customer_Satisfaction_Level'] = pd.cut(
    df_clean['Feedback_Score'],
    bins=[0, 3, 6, 8, 10],
    labels=['Dissatisfied', 'Neutral', 'Satisfied', 'Highly Satisfied']
)

# Recreate Customer Spending Segment
df_clean['Customer_Spending_Segment'] = pd.qcut(
    df_clean['Purchase_Amount'],
    q=4,
    labels=['Low', 'Moderate', 'High', 'Top']
)

# Recreate Customer Engagement Segment
df_clean['Customer_Engagement_Segment'] = pd.cut(
    df_clean['Days_Since_Last_Purchase'],
    bins=[-1, 30, 90, 180, float('inf')],
    labels=[
        'Recently Active',
        'Moderately Active',
        'Less Active',
        'Inactive'
    ]
)


# ==========================================
# 3. DATE-BASED COLUMNS
# ==========================================

# Recalculate Signup Month and Year
df_clean['Signup_Month'] = df_clean['Signup_Date'].dt.month
df_clean['Signup_Year'] = df_clean['Signup_Date'].dt.year

# Recalculate Last Purchase Month and Year
df_clean['Last_Purchase_Month'] = (
    df_clean['Last_Purchase_Date'].dt.month
)

df_clean['Last_Purchase_Year'] = (
    df_clean['Last_Purchase_Date'].dt.year
)


# ==========================================
# DISPLAY UPDATED DERIVED COLUMNS
# ==========================================

derived_columns = [
    'Purchase_Duration_Days',
    'Days_Since_Last_Purchase',
    'Purchase_Amount_Percentile',
    'Purchase_Amount_Difference_From_City_Median',
    'Age_Group',
    'Customer_Satisfaction_Level',
    'Customer_Spending_Segment',
    'Customer_Engagement_Segment',
    'Signup_Month',
    'Signup_Year',
    'Last_Purchase_Month',
    'Last_Purchase_Year'
]

display(df_clean[derived_columns].head())

   
   
```

After outlier treatment, **all 12 derived columns, including numerical, categorical, and date-based features, were recalculated using the updated dataset**. This ensures that the **derived values and customer classifications** reflect the final cleaned records. The updated columns provide a consistent basis for further **filtering, grouping, aggregation, and exploratory analysis.**
